```python
# Step 1: Load the clustered data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Load the data with cluster assignments
rfm_data = pd.read_csv('../data/processed/rfm_with_clusters.csv')

print(f"Loaded {len(rfm_data)} customers with cluster assignments")
print(f"Clusters: {sorted(rfm_data['Cluster'].unique())}")

rfm_data.head()

# Visualize clusters using PCA

X = rfm_data[['Recency', 'Frequency', 'Monetary']].values

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.figure(figsize=(10, 7))
colors = ['red', 'blue', 'green', 'orange', 'purple']

for cluster in sorted(rfm_data['Cluster'].unique()):
    mask = rfm_data['Cluster'] == cluster
    plt.scatter(X_pca[mask, 0], X_pca[mask, 1], 
               label=f'Cluster {cluster}',
               alpha=0.6, s=50,
               color=colors[cluster])

plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.1%})')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.1%})')
plt.title('Customer Clusters')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Step 2: Visualize clusters in 2D space using PCA
# PCA reduces 3 dimensions (R, F, M) to 2 for easy visualization

# Prepare data for PCA
X = rfm_data[['Recency', 'Frequency', 'Monetary']].values

# Apply PCA to get 2D representation
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Create scatter plot
plt.figure(figsize=(10, 7))
colors = ['red', 'blue', 'green', 'orange', 'purple']
clusters = rfm_data['Cluster'].unique()

for cluster in sorted(clusters):
    mask = rfm_data['Cluster'] == cluster
    plt.scatter(X_pca[mask, 0], X_pca[mask, 1], 
               label=f'Cluster {cluster}', 
               alpha=0.6, 
               s=50,
               color=colors[cluster])

plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.1%} variance)')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.1%} variance)')
plt.title('Customer Segments Visualization')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Analyze each cluster

for cluster in sorted(rfm_data['Cluster'].unique()):
    data = rfm_data[rfm_data['Cluster'] == cluster]
    print(f"\nCluster {cluster}")
    print(f"  Size: {len(data)} customers")
    print(f"  Avg Recency: {data['Recency'].mean():.0f} days")
    print(f"  Avg Frequency: {data['Frequency'].mean():.1f} orders")
    print(f"  Avg Monetary: £{data['Monetary'].mean():.2f}")
```

```python
# Step 3: Analyze each cluster's characteristics

print("="*60)
print("CLUSTER ANALYSIS - WHO ARE THESE CUSTOMERS?")
print("="*60)

for cluster in sorted(rfm_data['Cluster'].unique()):
    cluster_data = rfm_data[rfm_data['Cluster'] == cluster]
    
    print(f"\n📊 CLUSTER {cluster} ({len(cluster_data)} customers)")
    print("-" * 60)
    print(f"  Recency (avg days since purchase): {cluster_data['Recency'].mean():.0f}")
    print(f"  Frequency (avg # of orders): {cluster_data['Frequency'].mean():.1f}")
    print(f"  Monetary (avg spend): £{cluster_data['Monetary'].mean():.2f}")
    print(f"  Total value: £{cluster_data['Monetary'].sum():.2f}")
    print(f"  % of customer base: {(len(cluster_data)/len(rfm_data)*100):.1f}%")
```

```python
# Step 4: Visualize cluster characteristics

# Compare clusters visually

stats = rfm_data.groupby('Cluster')[['Recency', 'Frequency', 'Monetary']].mean()

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

stats['Recency'].plot(kind='bar', ax=axes[0], color='coral')
axes[0].set_title('Avg Recency')
axes[0].set_ylabel('Days')

stats['Frequency'].plot(kind='bar', ax=axes[1], color='skyblue')
axes[1].set_title('Avg Frequency')
axes[1].set_ylabel('Orders')

stats['Monetary'].plot(kind='bar', ax=axes[2], color='lightgreen')
axes[2].set_title('Avg Monetary')
axes[2].set_ylabel('£')

plt.tight_layout()
plt.show()
```

```python
# Step 5: Create customer segment profiles

print("\n" + "="*70)
print("CUSTOMER SEGMENT PROFILES & MARKETING RECOMMENDATIONS")
print("="*70)

# Define profiles based on typical RFM patterns
profiles = {
    0: {
        'name': 'At-Risk Customers',
        'description': 'Haven\'t purchased recently but were once valuable',
        'recommendation': 'Win-back campaigns, special offers'
    },
    1: {
        'name': 'Loyal Customers',
        'description': 'Regular buyers with consistent spending',
        'recommendation': 'VIP programs, exclusive benefits'
    },
    2: {
        'name': 'High-Value Customers',
        'description': 'Spend the most money, frequent purchasers',
        'recommendation': 'Premium tier, personalized service'
    },
    3: {
        'name': 'Potential Customers',
        'description': 'New or occasional buyers with growth potential',
        'recommendation': 'Engagement campaigns, upselling'
    }
}

# Create segment profiles

for cluster in sorted(rfm_data['Cluster'].unique()):
    data = rfm_data[rfm_data['Cluster'] == cluster]
    
    avg_r = data['Recency'].mean()
    avg_f = data['Frequency'].mean()
    avg_m = data['Monetary'].mean()
    
    if avg_f > rfm_data['Frequency'].median() and avg_m > rfm_data['Monetary'].median():
        name = "Champions"
    elif avg_r < rfm_data['Recency'].median():
        name = "Active"
    elif avg_m < rfm_data['Monetary'].median():
        name = "Budget"
    else:
        name = "Regular"
    
    print(f"\nCluster {cluster}: {name}")
    print(f"  {len(data)} customers")
    print(f"  Value: £{avg_m:.2f}")

# Show profiles (adjust based on actual cluster count)
for cluster in sorted(rfm_data['Cluster'].unique()):
    if cluster < len(profiles):
        profile = profiles.get(cluster, {'name': f'Segment {cluster}', 'description': 'TBD', 'recommendation': 'TBD'})
        print(f"\n🎯 CLUSTER {cluster}: {profile['name']}")
        print(f"   Description: {profile['description']}")
        print(f"   Action: {profile['recommendation']}")

print("\n✓ Visualization complete! Ready for the final report.")
```
