# Phase 2: Data Cleaning & Preprocessing

This notebook focuses on cleaning and preparing the raw data for exploratory data analysis and feature engineering.

**Goals:**
- Remove missing and invalid records
- Fix data types
- Clean text fields
- Create useful derived columns
- Save cleaned dataset for downstream analysis

**Tracking Milestone:**
✔ Remove missing CustomerIDs
✔ Remove negative quantities
✔ Remove cancelled transactions
✔ Fix data types
✔ Clean description text
✔ Save cleaned dataset to `data/processed/`

```python
# Cell 1: Imports & Load Data

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Load the raw data
file_path = "../data/raw/Online Retail.csv"
df = pd.read_csv(file_path, encoding='ISO-8859-1')

print(f"✓ Loaded {len(df):,} rows from raw data")
print(f"Initial shape: {df.shape}")
print(f"\nColumns: {list(df.columns)}")
print(f"\nFirst few rows:")
df.head()
```

```python
# Cell 2: Initial Data Assessment

print("=" * 60)
print("INITIAL DATA QUALITY ASSESSMENT")
print("=" * 60)

print("\n📊 Missing Values:")
missing = df.isnull().sum()
print(missing[missing > 0])

print("\n📋 Data Types:")
print(df.dtypes)

print("\n🔢 Unique Values:")
print(f"  • Invoices: {df['InvoiceNo'].nunique():,}")
print(f"  • Customers: {df['CustomerID'].nunique():,}")
print(f"  • Products: {df['StockCode'].nunique():,}")
print(f"  • Countries: {df['Country'].nunique():,}")

print("\n💰 Quantity & Price Statistics:")
print(f"  • Min Quantity: {df['Quantity'].min()}")
print(f"  • Max Quantity: {df['Quantity'].max()}")
print(f"  • Min Price: £{df['UnitPrice'].min():.2f}")
print(f"  • Max Price: £{df['UnitPrice'].max():.2f}")
```

```python
# Cell 3: Remove Missing CustomerIDs

print("\n" + "=" * 60)
print("STEP 1: Remove Missing CustomerIDs")
print("=" * 60)

rows_before = len(df)
df = df.dropna(subset=['CustomerID'])
rows_removed = rows_before - len(df)

print(f"Rows before: {rows_before:,}")
print(f"Rows removed: {rows_removed:,}")
print(f"Rows after: {len(df):,}")
print(f"Progress: {len(df)/rows_before*100:.1f}% of data retained")
```

```python
# Cell 4: Remove Negative or Zero Quantities

print("\n" + "=" * 60)
print("STEP 2: Remove Negative or Zero Quantities")
print("=" * 60)

# Check for negative/zero quantities
print(f"\nNegative quantities: {len(df[df['Quantity'] < 0])}")
print(f"Zero quantities: {len(df[df['Quantity'] == 0])}")

rows_before = len(df)
df = df[df['Quantity'] > 0]
rows_removed = rows_before - len(df)

print(f"\nRows before: {rows_before:,}")
print(f"Rows removed: {rows_removed:,}")
print(f"Rows after: {len(df):,}")
print(f"Progress: {len(df)/rows_before*100:.1f}% of data retained")
```

```python
# Cell 5: Remove Cancelled Transactions

print("\n" + "=" * 60)
print("STEP 3: Remove Cancelled Transactions")
print("=" * 60)

# Cancelled invoices start with 'C'
cancelled_count = len(df[df['InvoiceNo'].astype(str).str.startswith('C')])
print(f"Cancelled transactions (InvoiceNo starts with 'C'): {cancelled_count:,}")

rows_before = len(df)
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
rows_removed = rows_before - len(df)

print(f"\nRows before: {rows_before:,}")
print(f"Rows removed: {rows_removed:,}")
print(f"Rows after: {len(df):,}")
print(f"Progress: {len(df)/rows_before*100:.1f}% of data retained")
```

```python
# Cell 6: Remove Negative or Zero Prices

print("\n" + "=" * 60)
print("STEP 4: Remove Negative or Zero Prices")
print("=" * 60)

# Check for negative/zero prices
print(f"Negative prices: {len(df[df['UnitPrice'] < 0])}")
print(f"Zero prices: {len(df[df['UnitPrice'] == 0])}")

rows_before = len(df)
df = df[df['UnitPrice'] > 0]
rows_removed = rows_before - len(df)

print(f"\nRows before: {rows_before:,}")
print(f"Rows removed: {rows_removed:,}")
print(f"Rows after: {len(df):,}")
print(f"Progress: {len(df)/rows_before*100:.1f}% of data retained")
```

```python
# Cell 7: Fix Data Types

print("\n" + "=" * 60)
print("STEP 5: Fix Data Types")
print("=" * 60)

# Convert InvoiceDate to datetime
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Convert CustomerID to integer
df['CustomerID'] = df['CustomerID'].astype(int)

# Ensure Quantity is integer
df['Quantity'] = df['Quantity'].astype(int)

print("Data types fixed:")
print(f"  • InvoiceDate → datetime64")
print(f"  • CustomerID → int64")
print(f"  • Quantity → int64")

print("\n✓ Updated Data Types:")
print(df.dtypes)
```

```python
# Cell 8: Clean Description Text

print("\n" + "=" * 60)
print("STEP 6: Clean Description Text")
print("=" * 60)

# Clean: strip whitespace and convert to lowercase
df['Description'] = df['Description'].str.strip().str.lower()

print("Description cleaned:")
print(f"  • Removed leading/trailing whitespace")
print(f"  • Converted to lowercase")

print("\nSample cleaned descriptions:")
print(df['Description'].head(10).values)
```

```python
# Cell 9: Create Total Price Column

print("\n" + "=" * 60)
print("STEP 7: Create Total Price Column")
print("=" * 60)

# Calculate total price per transaction
df['Amount'] = df['Quantity'] * df['UnitPrice']

print("Amount column created: Quantity × UnitPrice")

print("\nAmount Statistics:")
print(f"  • Min Amount: £{df['Amount'].min():.2f}")
print(f"  • Max Amount: £{df['Amount'].max():.2f}")
print(f"  • Mean Amount: £{df['Amount'].mean():.2f}")
print(f"  • Median Amount: £{df['Amount'].median():.2f}")

print("\nSample data with Amount column:")
df[['InvoiceNo', 'Quantity', 'UnitPrice', 'Amount']].head(10)
```

```python
# Cell 10: Check Final Data Quality

print("\n" + "=" * 60)
print("STEP 8: Final Data Quality Check")
print("=" * 60)

print("\n📊 Final Missing Values:")
missing = df.isnull().sum()
if missing.sum() == 0:
    print("  ✓ No missing values detected!")
else:
    print(missing[missing > 0])

print("\n📋 Final Data Types:")
print(df.dtypes)

print("\n🔢 Final Dataset Summary:")
print(f"  • Total Rows: {len(df):,}")
print(f"  • Total Columns: {len(df.columns)}")
print(f"  • Unique Invoices: {df['InvoiceNo'].nunique():,}")
print(f"  • Unique Customers: {df['CustomerID'].nunique():,}")
print(f"  • Unique Products: {df['StockCode'].nunique():,}")
print(f"  • Countries: {df['Country'].nunique():,}")

print("\n✅ Data Quality Assessment: PASSED")
print("   All missing values handled, data types corrected, invalid records removed.")
```

```python
# Cell 11: Save Cleaned Dataset

print("\n" + "=" * 60)
print("STEP 9: Save Cleaned Dataset")
print("=" * 60)

# Save to CSV
output_path = "../data/processed/online_retail_cleaned.csv"
df.to_csv(output_path, index=False)

print(f"✓ Cleaned dataset saved to: {output_path}")
print(f"  • Rows: {len(df):,}")
print(f"  • Columns: {len(df.columns)}")
print(f"  • File size: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

# Display final cleaned data sample
print("\n✓ Final Cleaned Data Sample:")
print(df.head(10))
```
